```python
## Task 1 ##

## Given ##
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
                     [2, 1, 3],
                     [1, 0, 2],
                     [4, 1, 1] ])
print(matrix_a)
print("\n")
matrix_b = np.array([
[1, 2, 0],
[3, 1, 4],
[2, 5, 1]
])

print(matrix_b)
```

```python
## Vector ##

addition = vector_a + vector_b
print("Vector_Addition:",addition)
print("\n")
subtraction = vector_a - vector_b
print("Vector_subtraction:",subtraction)
print("\n")
dot_product = np.dot(vector_a, vector_b)
print("Vector_dot_product:",dot_product)
print("\n")
```

```python
## Matrix ##
matrix_addition = matrix_a + matrix_b


matrix_multiplication = np.dot(matrix_a, matrix_b)

print("Matrix Addition:")
print(matrix_addition)
print("\n")

print("Matrix Multiplication:")
print(matrix_multiplication)
```

```python
determinant = np.linalg.det(matrix_a)
print("Determinant:", determinant)
```

```python
if determinant!= 0:
  rev_matrix = np.linalg.inv(matrix_a)
  print("Inverse of matrix_a:",rev_matrix)
else:
    print("Inverse does not exist")
```

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("Eigenvalues:", eigenvalues)
print("\n")
print("Eigenvectors:", eigenvectors)
```

```python
## Binomial Distribution ##

## P(X=k)=(n/k)p^k * (1-p)^(n-k)

n = 20
p = 0.65
for k in range(n + 1):
    probability = binom.pmf(k, n, p)
    print(f"k = {k}, Probability = {probability:.3f}")

```

```python
## normal distribution ##

normal_mean = 50
normal_std = 10

normal_distri = np.linspace(40, 100, 100)
y = norm.pdf(normal_distri, normal_mean, normal_std)

plt.plot(normal_distri, y)
plt.title("Normal Distribution")
plt.show()
```

Interpretation of the Normal Distribution Graph:

The graph represents a normal distribution with a mean of 50 and a standard deviation of 10.

1. The curve is bell-shaped and symmetrical around the mean (50).
2. The highest point of the curve is at 50, indicating that values around 50 have the highest probability density.
3. The standard deviation of 10 indicates how widely the values are spread around the mean.
4. Most observations are expected to fall close to the mean.
5. Values farther away from 50 have lower probability density.
6. The curve gradually approaches the X-axis on both sides.

Important statistical interpretation
According to the Empirical Rule:

Range	Percentage of data

40–60 (Mean ± 1 SD)	68.27%
30–70 (Mean ± 2 SD)	95.45%
20–80 (Mean ± 3 SD)	99.73%

Conclusion:
The dataset follows a normal distribution pattern, where most values are concentrated around 50, and fewer values occur as we move farther away from the mean.

```python
## uniform distribution ##
uniform_low = 0
uniform_high = 1

## np.random.uniform(low, high, size = optional) ##
uniform_distribution = np.random.uniform(0,1)
print("uniform_distribution =  " ,uniform_distribution)

plt.hist(uniform_distribution, bins=3, edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

Interpretation of Uniform Distribution:

The code generates a random number between 0 and 1 using np.random.uniform() and displays it using a histogram.

Interpretation:
1. A random number is generated between 0 and 1.
2. In a uniform distribution, every value within the specified range has an equal probability of being selected.
3. There is no particular value that is more likely to occur than another.
4. The distribution is expected to be evenly spread when a large number of random values are generated.
5. The histogram represents the frequency of values in different intervals.
Important observation :

uniform_distribution = np.random.uniform(0, 1)

Here, only one random number is generated.

Therefore, your histogram will not show a meaningful uniform distribution.

Improved code:

To understand uniform distribution better, generate 1,000 random numbers.

uniform_distribution = np.random.uniform(0, 1, 1000)

plt.hist(uniform_distribution, bins=10,edgecolor="black")

Conclusion:  
A uniform distribution means that all values within a given range have an equal chance of occurring. When a large number of random values are generated, the histogram generally shows approximately equal frequencies across the intervals.

```python
## Task - 3 ##
## Descriptive Statistics ##

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

mean = np.mean(sales_data)
median = np.median(sales_data)
mode = statistics.mode(sales_data)

print("Mean = ",mean)
print("Median = ",median)
print("Mode = ",mode)

```

```python
## variance, Range and Standard deviation ##

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

Data_Set = pd.Series(sales_data)
Range_Data_Set = Data_Set.max()-Data_Set.min()
print("Range = ",Range_Data_Set)

Variance_Data_Set = Data_Set.var(ddof=0)
print("Variance = ",Variance_Data_Set)

Std_Data_Set = Data_Set.std(ddof=0)
print("Standard_Deviation = ",Std_Data_Set)
```

```python
## Skewness, kurtosis ##

Skewness = Data_Set.skew()
print("Skewness =", Skewness)
kurtosis = Data_Set.kurt()
print("Kurtosis = ",kurtosis)
```

## Interpretation ##

Is the dataset symmetric or skewed?

Our skewness is:

Skewness = 0.033


- Positive value means positive skewness.
- Negative value means negative skewness.
- Zero means perfectly symmetrical.

Since 0.033 is very close to zero, our sales dataset is approximately symmetrical, with a very slight positive skew.


Our kurtosis is:

Kurtosis = -0.549


Since it is negative, the distribution is classified as platykurtic (using excess kurtosis).

This indicates lighter tails than a normal distribution.

Conclusion:

The sales dataset has a mean of 165.5 and a median of 165.

The skewness value is 0.033, indicating that the distribution is approximately symmetrical with a very slight positive skew.

The kurtosis value is -0.549, indicating a platykurtic distribution with lighter tails than a normal distribution.

Therefore, the sales data is nearly symmetrical.
